Example

Input:

Alice private key: 97
Bob private key: 233

Output:

Alice Public Key: 40
Bob Public Key: 248

Alice Secret Key: 160
Bob Secret Key: 160

দুইজন একই secret পেয়ে গেল।

```python
q = 353
a = 3

Xa = int(input("Alice private key: "))
Xb = int(input("Bob private key: "))


# Public keys
Xa = pow(a, Xa, q)
Xb = pow(a, Xb, q)


# Shared secret
key_A = pow(Xb, Xa, q)
key_B = pow(Xa, Xb, q)


print("Alice Public Key:", Xa)
print("Bob Public Key  :", Xb)

print("Alice Secret Key:", key_A)
print("Bob Secret Key  :", key_B)
```
